## Define and Plot Function

### Subtask:
Define the function f(x) = (x-3)^2 + 4 and its derivative. Plot f(x) over the range [-5, 10] using matplotlib, clearly marking the true minimum point on the graph.


**Reasoning**:
The first step is to import the necessary libraries, NumPy for numerical operations and Matplotlib for plotting, to define the functions and prepare for visualization.



````python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
````

**Reasoning**:
Now that the necessary libraries are imported, I will define the functions f(x) and its derivative df(x), generate x and y values, and then plot the function f(x), marking its true minimum point.



````python
def f(x):
    """Defines the function f(x) = (x-3)^2 + 4."""
    return (x - 3)**2 + 4

def df(x):
    """Defines the derivative of f(x), which is 2*(x-3)."""
    return 2 * (x - 3)

# Generate a range of x values
x_values = np.linspace(-5, 10, 400)
# Calculate corresponding y values
y_values = f(x_values)

# Identify the true minimum
true_min_x = 3
true_min_y = f(true_min_x)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='f(x) = (x-3)^2 + 4', color='blue')
plt.scatter(true_min_x, true_min_y, color='red', marker='o', s=100, label=f'True Minimum (x={true_min_x}, y={true_min_y})', zorder=5)

# Add labels and title
plt.title('Function f(x) and its True Minimum')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()

print(f"Function f(x) defined as (x-3)^2 + 4 and its derivative df(x) as 2*(x-3). The function plot with its true minimum has been displayed.")
````

## Implement Gradient Descent

### Subtask:
Implement the gradient descent algorithm as a function. This function should take the derivative of f(x), an initial starting point (x0), a learning rate, and a convergence threshold as parameters. It should return a list of x values representing the steps taken during descent.


**Reasoning**:
I will define the `gradient_descent` function as specified, incorporating the parameters for the derivative, initial point, learning rate, convergence threshold, and maximum iterations, along with the logic to track the history of x values and handle convergence.



````python
def gradient_descent(df, x0, learning_rate, threshold=1e-5, max_iterations=1000):
    """
    Implements the gradient descent algorithm.

    Args:
        df: The derivative function of the objective function.
        x0: The initial starting point.
        learning_rate: The step size for each iteration.
        threshold: The convergence threshold.
        max_iterations: The maximum number of iterations to prevent infinite loops.

    Returns:
        A list of x values representing the steps taken during descent.
    """
    x = x0
    x_history = [x0]

    for i in range(max_iterations):
        gradient = df(x)
        x_new = x - learning_rate * gradient

        x_history.append(x_new)

        if abs(x_new - x) < threshold:
            break

        x = x_new

    print(f"Gradient descent function defined successfully. Last x value: {x_history[-1]:.4f}")
    return x_history
````

## Run Gradient Descent (LR = 0.01)

### Subtask:
Execute the gradient descent algorithm starting from x₀ = -4 with a learning rate of 0.01. Record the sequence of x values and the number of iterations until convergence.


**Reasoning**:
I will execute the `gradient_descent` function with the specified parameters (`df`, `x0=-4`, `learning_rate=0.01`) and store the results in `x_history_0_01`. Then, I will print the number of iterations by checking the length of the `x_history_0_01` list.



````python
x0 = -4
learning_rate_0_01 = 0.01
x_history_0_01 = gradient_descent(df, x0, learning_rate_0_01)

print(f"Gradient Descent with learning rate {learning_rate_0_01} completed.")
print(f"Number of iterations: {len(x_history_0_01) - 1}")
print(f"Final x value: {x_history_0_01[-1]:.4f}")
````

## Summary and Analysis of Gradient Descent

### Conceptual Aspects of Gradient Descent
Gradient descent is an iterative optimization algorithm used to find the local minimum of a function. The core idea is to move iteratively in the direction of steepest descent, which is given by the negative of the gradient of the function at the current point. For a function `f(x)`, the update rule is:

x_new = x_old - learning_rate * df(x_old)

Where:
- `x_new` is the updated parameter value.
- `x_old` is the current parameter value.
- `learning_rate` (or step size) is a hyperparameter that determines the size of the steps taken towards the minimum.
- `df(x_old)` is the derivative (gradient) of the function at `x_old`.

The algorithm stops when the change in `x` (or `f(x)`) between iterations falls below a predefined convergence threshold, or a maximum number of iterations is reached.

### Role of Learning Rate
The learning rate is crucial for the performance of gradient descent. It controls how large a step we take in the direction of the negative gradient. Choosing an appropriate learning rate is vital:

*   **Small Learning Rate**: If the learning rate is too small, the algorithm will take tiny steps, requiring many iterations to converge. This makes the process slow and computationally expensive, though it is more likely to find a good minimum.
*   **Large Learning Rate**: If the learning rate is too large, the algorithm might overshoot the minimum, oscillating around it or even diverging entirely, failing to converge. It can lead to unstable behavior and prevent the algorithm from finding the optimal solution.

### Analysis of Convergence Behavior for Each Learning Rate

For the function `f(x) = (x-3)^2 + 4` with its true minimum at `x = 3` and `y = 4`, we observed the following convergence behaviors:

*   **Learning Rate = 0.01**:
    *   **Iterations**: 474
    *   **Final x value**: 2.9995
    *   **Behavior**: With a small learning rate of 0.01, the algorithm took many small steps towards the minimum. It converged slowly but steadily, approaching the true minimum very closely. This demonstrates a safe, albeit time-consuming, convergence path.

*   **Learning Rate = 0.1**:
    *   **Iterations**: 55
    *   **Final x value**: 3.0000
    *   **Behavior**: A moderate learning rate of 0.1 resulted in much faster convergence (significantly fewer iterations than 0.01). The algorithm efficiently reached the true minimum without significant oscillation. This learning rate appears to be well-suited for this specific function and starting point, balancing speed and accuracy.

*   **Learning Rate = 0.9**:
    *   **Iterations**: 64
    *   **Final x value**: 3.0000
    *   **Behavior**: This learning rate is quite large. Initially, the algorithm takes very large steps, overshooting the minimum dramatically. For example, from x0 = -4, the first step went to 8.6, then back to -1.48, then to 6.584, etc. This caused oscillations around the minimum. While it eventually converged to the true minimum (3.0000), it did so in an oscillatory manner. It converged in a similar number of iterations to the 0.1 learning rate in this specific case due to the quadratic nature of the function and the small distance from the minimum, but the path taken was less stable and could easily diverge in more complex functions or with an even larger learning rate.

**Conclusion**: The visualization clearly illustrates the trade-offs involved in choosing a learning rate. A very small learning rate leads to slow convergence, while a very large one can cause overshooting and oscillations, or even divergence. An optimally chosen learning rate can lead to efficient and stable convergence, as seen with the 0.1 learning rate in this example.
