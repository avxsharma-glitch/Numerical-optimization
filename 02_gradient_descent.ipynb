# Task
Implement and analyze the Gradient Descent algorithm for linear regression using the provided `hours_studied` and `exam_scores` datasets, including data normalization, tracking Mean Squared Error, visualizing the best-fit line and learning curve, and comparing the effects of different learning rates (0.01, 0.5, and 0.0001) on the model's convergence and performance.

## Prepare Data

### Subtask:
Define the given `hours_studied` and `exam_scores` datasets. Normalize the `hours_studied` data to improve the stability of Gradient Descent.


**Reasoning**:
First, I'll define the `hours_studied` and `exam_scores` datasets as NumPy arrays. Then, I'll apply Min-Max normalization to the `hours_studied` data and store it in a new variable `normalized_hours_studied`. Finally, I'll display the original and normalized data to verify the transformation.



````python
import numpy as np

# Define the hours_studied and exam_scores datasets
hours_studied = np.array([10, 9, 2, 15, 10, 16, 11, 16, 9, 10])
exam_scores = np.array([95, 80, 10, 80, 70, 90, 70, 75, 60, 50])

# Normalize the hours_studied data using Min-Max scaling
min_hours = np.min(hours_studied)
max_hours = np.max(hours_studied)

normalized_hours_studied = (hours_studied - min_hours) / (max_hours - min_hours)

print("Original hours_studied:", hours_studied)
print("Normalized hours_studied:", normalized_hours_studied)
print("Exam scores:", exam_scores)
````

## Implement Gradient Descent

### Subtask:
Implement the Gradient Descent algorithm for linear regression. Initialize `m` (slope) and `c` (intercept) to 0. Set `learning_rate = 0.01` and `iterations = 1000`. Predict exam scores, compute errors, calculate gradients, and update `m` and `c` within a loop. Track the Mean Squared Error (MSE) in a list called `cost_history` for each iteration.


**Reasoning**:
I will implement the Gradient Descent algorithm following the provided instructions, including initialization of parameters, iterative calculation of predictions, errors, gradients, and parameter updates, and tracking the Mean Squared Error in `cost_history`.



````python
m = 0
c = 0
learning_rate = 0.01
iterations = 1000
n = len(normalized_hours_studied)

cost_history = []

for i in range(iterations):
    y_pred = m * normalized_hours_studied + c
    error = y_pred - exam_scores

    grad_m = (1/n) * np.sum(error * normalized_hours_studied)
    grad_c = (1/n) * np.sum(error)

    m = m - learning_rate * grad_m
    c = c - learning_rate * grad_c

    mse = (1/n) * np.sum(error**2)
    cost_history.append(mse)

print(f"Final optimized m: {m}")
print(f"Final optimized c: {c}")
print(f"Last calculated MSE: {cost_history[-1]}")
````

## Visualize Best-Fit Line

### Subtask:
Create a scatter plot of the original `hours_studied` vs `exam_scores`. On the same plot, overlay the best-fit line learned by Gradient Descent using the final optimized values of `m` and `c`. Ensure the plot has appropriate labels and a legend.


**Reasoning**:
To visualize the best-fit line, I need to import `matplotlib.pyplot`, plot the scatter points of the original data, then plot the regression line using the optimized `m` and `c` values, and finally add appropriate labels, title, and a legend.



````python
import matplotlib.pyplot as plt

# Calculate predicted y values for the best-fit line
y_values = m * normalized_hours_studied + c

# Create the plot
plt.figure(figsize=(10, 6))
plt.scatter(normalized_hours_studied, exam_scores, label='Original Data')
plt.plot(normalized_hours_studied, y_values, color='red', label='Best-Fit Line')

# Add labels and title
plt.title('Best-Fit Line from Gradient Descent')
plt.xlabel('Normalized Hours Studied')
plt.ylabel('Exam Scores')
plt.legend()
plt.grid(True)
plt.show()
````

## Plot Learning Curve

### Subtask:
Generate a line plot of `cost_history` (MSE) against the number of iterations. This plot will show how the error decreases as training progresses, demonstrating the learning process of the Gradient Descent algorithm. Ensure the plot has appropriate labels and a legend.


**Reasoning**:
I need to generate a line plot of `cost_history` against the number of iterations to visualize the learning curve of the Gradient Descent algorithm. I will use `matplotlib.pyplot` for this.



````python
plt.figure(figsize=(10, 6))
plt.plot(range(iterations), cost_history, label='MSE Cost')
plt.title('Gradient Descent Learning Curve (MSE vs. Iterations)')
plt.xlabel('Iterations')
plt.ylabel('Mean Squared Error (MSE)')
plt.legend()
plt.grid(True)
plt.show()
````

## Bonus Challenge: High Learning Rate

### Subtask:
Re-run the Gradient Descent implementation with a `learning_rate = 0.5`. Store the `m`, `c`, and `cost_history` for this run. Visualize the new best-fit line and the cost history curve.


**Reasoning**:
I will implement the Gradient Descent algorithm with a learning rate of 0.5, track the `m`, `c`, and `cost_history`, and then visualize the best-fit line and learning curve. I will use the instructions provided to initialize variables and perform the calculations within a loop. Afterwards, I will use matplotlib to generate the required plots.



````python
m_high_lr = 0
c_high_lr = 0
learning_rate_high = 0.5
iterations = 1000
n = len(normalized_hours_studied)

cost_history_high_lr = []

for i in range(iterations):
    y_pred_high_lr = m_high_lr * normalized_hours_studied + c_high_lr
    error_high_lr = y_pred_high_lr - exam_scores

    grad_m_high_lr = (1/n) * np.sum(error_high_lr * normalized_hours_studied)
    grad_c_high_lr = (1/n) * np.sum(error_high_lr)

    m_high_lr = m_high_lr - learning_rate_high * grad_m_high_lr
    c_high_lr = c_high_lr - learning_rate_high * grad_c_high_lr

    mse_high_lr = (1/n) * np.sum(error_high_lr**2)
    cost_history_high_lr.append(mse_high_lr)

print(f"Final optimized m_high_lr: {m_high_lr}")
print(f"Final optimized c_high_lr: {c_high_lr}")
print(f"Last calculated MSE (high LR): {cost_history_high_lr[-1]}")

# Calculate predicted y values for the best-fit line with high LR
y_values_high_lr = m_high_lr * normalized_hours_studied + c_high_lr

# Create the plot for best-fit line
plt.figure(figsize=(10, 6))
plt.scatter(normalized_hours_studied, exam_scores, label='Original Data')
plt.plot(normalized_hours_studied, y_values_high_lr, color='red', label='Best-Fit Line (LR=0.5)')

# Add labels and title
plt.title('Best-Fit Line from Gradient Descent (Learning Rate = 0.5)')
plt.xlabel('Normalized Hours Studied')
plt.ylabel('Exam Scores')
plt.legend()
plt.grid(True)
plt.show()

# Create the plot for learning curve
plt.figure(figsize=(10, 6))
plt.plot(range(iterations), cost_history_high_lr, label='MSE Cost (LR=0.5)')
plt.title('Gradient Descent Learning Curve (Learning Rate = 0.5)')
plt.xlabel('Iterations')
plt.ylabel('Mean Squared Error (MSE)')
plt.legend()
plt.grid(True)
plt.show()
````

**Reasoning**:

Now, to complete the comparison aspect of the main task, I will implement Gradient Descent with a very low learning rate (0.0001), store its parameters and cost history, and then visualize its best-fit line and learning curve.
